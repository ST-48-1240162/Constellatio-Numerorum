# Random stable IIR pole cloud on a Colab GPU

Lattice reflection coefficients with `|k|<1` (Schur). Step-up gives direct-form poles inside the unit circle. Color is filter order. Brightness is pole Q from the 3 dB bandwidth.

1. Runtime → Change runtime type → **T4 GPU** or **A100**.
2. Runtime → Run all.

The first cell clones this repo if the tree is missing. Default is orders 2–18, about 1500 filters on the high end, 3840×2160.

```python
import os
from pathlib import Path

try:
    import google.colab
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

if IN_COLAB and not Path("iir-poles/lattice_poles.py").exists():
    !git clone --depth 1 https://github.com/ST-48-1240162/Constellatio-Numerorum.git /content/Constellatio-Numerorum
    os.chdir("/content/Constellatio-Numerorum")

import torch
print("cwd", Path.cwd())
print("torch", torch.__version__)
print("cuda", torch.cuda.is_available(), end=" ")
if torch.cuda.is_available():
    print(torch.cuda.get_device_name(0))
else:
    print("(CPU fallback)")
```

```python
import runpy
import sys
from pathlib import Path

out = Path("iir-poles") / "poles_o18_3840px.png"
sys.argv = [
    "lattice_poles.py",
    "--orders", "2-18",
    "--n", "1500",
    "--width", "3840",
    "--height", "2160",
    "--gpu",
    "--batch", "4096",
    "--png", str(out),
]
runpy.run_path(str(Path("iir-poles") / "lattice_poles.py"), run_name="__main__")
print("wrote", out)
```
