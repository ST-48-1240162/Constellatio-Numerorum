# Algebraic plane scatter on a Colab GPU

Box enumeration: monic-ish polynomials with integer coeffs in `[-L, L]`, roots in the complex plane. Color is degree; brightness and size follow Brooks-style blob weighting.

1. Runtime → Change runtime type → **T4 GPU** (free) or **A100** (paid).
2. Runtime → Run all.

Default here: `coeff-range=4`, degrees 1–5, **3840×2160** (16:9 when x/y limits match). Expect ~2.5M roots and several minutes on a T4.

```python
import os
from pathlib import Path

try:
    import google.colab
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

REPO = Path("/content/constellatio-numerorum")
REPO_URL = "https://github.com/ST-48-1240162/constellatio-numerorum.git"

if IN_COLAB:
    if not REPO.exists():
        !git clone --depth 1 {REPO_URL} {REPO}
    else:
        !git -C {REPO} pull
    os.chdir(REPO)

import torch
print("cwd", Path.cwd())
print("torch", torch.__version__)
print("cuda", torch.cuda.is_available(), end=" ")
if torch.cuda.is_available():
    print(torch.cuda.get_device_name(0))
else:
    print("(enable a GPU runtime or this falls back to CPU)")
```

```python
import runpy
import sys
from pathlib import Path

WIDTH = 3840
COEFF_RANGE = 4
MAX_DEGREE = 5
BATCH = 8192

out = Path("algebraics") / f"algebraic_plane_{WIDTH}px_c{COEFF_RANGE}.png"
sys.argv = [
    "algebraic_plane.py",
    "--gpu",
    "--enum", "box",
    "--coeff-range", str(COEFF_RANGE),
    "--max-degree", str(MAX_DEGREE),
    "--width", str(WIDTH),
    "--batch", str(BATCH),
    "--png", str(out),
]
runpy.run_path(str(Path("algebraics") / "algebraic_plane.py"), run_name="__main__")
print("wrote", out)
```
